# converted 0 and 1 value to yes and no for better understanding 

```python
def conv(value):
    if value == 1:
        return "yes"
    else:
        return "no"
df['SeniorCitizen']= df['SeniorCitizen'].apply(conv)       
        
```

```python
ax = sns.countplot(x = 'Churn', data =df)
ax.bar_label(ax.containers[0])
plt.title("Count Of Customer by Churn")
plt.show
```

```python
gb = df.groupby("Churn").agg({'Churn':"count"})
gb
```

```python
plt.figure(figsize=(5,5))
plt.pie(gb['Churn'],labels = gb.index , autopct = "%1.2f%%" )
plt.title("Percentage  Of Customer by Churn")
plt.show
```

```python
plt.figure(figsize = (5,5))
sns.countplot(x = "gender",data =df, hue = "Churn")
plt.title("Chrun by Gender")
plt.show
```

```python
plt.figure(figsize = (5,5))
sns.countplot(x = "SeniorCitizen",data =df, hue = "Churn")
plt.title("Chrun by SeniorCitizen")
plt.show
```

```python
# stack bar chart
total_counts =df.groupby('SeniorCitizen')["Churn"].value_counts(normalize=True).unstack() *100
#plot
fig, ax = plt.subplots(figsize=(7,5)) 
#plot the bars
total_counts.plot(kind='bar',stacked  =True ,ax=ax,color=['#1f77b4','#ff7f0e'])

for p in ax.patches:
    width,height =p.get_width(),p.get_height()
    x,y =p.get_xy()
    ax.text(x +width /2, y + height /2 ,f'{height:.1f}%',ha='center',va='center')

plt.title('Churn by Senior Citizen ')
plt.xlabel('SeniorCitizen')
plt.ylabel('Percentage (%)')
plt.xticks(rotation = 0)
plt.legend(title='Churn',loc='upper right')
plt.show()
```

```python
df.columns.values
```

```python

columns = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

n_cols = 3
n_rows = (len(columns) + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, n_rows * 4))
axes = axes.flatten()

for i, col in enumerate(columns):
    sns.countplot(x=col, data=df, ax=axes[i] ,hue =df["Churn"])
    axes[i].set_title(f'Count Plot of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Count')
    axes[i].tick_params(axis='x', rotation=30)

# Remove empty subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

# This visualization presents count plots for key telecom services, showing how many customers use or do not use each service. It helps quickly identify the most and least adopted services across the customer base. The grid layout enables easy comparison between different service features. Overall, it supports exploratory data analysis by revealing usage patterns and potential drivers of customer behavior.
